## Data frame kurmak

Her sütun bir vektör. `data.frame()` bunları yan yana koyuyor.

```r
ogrenciler <- data.frame(
  isim   = c("Ayşe", "Mehmet", "Zeynep", "Can", "Elif", "Burak"),
  bolum  = c("Bilgisayar", "Endüstri", "Bilgisayar", "Elektrik", "Endüstri", "Bilgisayar"),
  vize   = c(72, 55, 88, 61, 90, 47),
  final  = c(80, 62, 95, 58, 85, 70)
)
ogrenciler
```

## Tabloyu tanımak

```r
str(ogrenciler)
nrow(ogrenciler)
ncol(ogrenciler)
```

## Bir sütuna ulaşmak: $

```r
ogrenciler$vize
mean(ogrenciler$final)
```

## Yeni sütun eklemek

Vize yüzde 40, final yüzde 60 ağırlıklı.

```r
ogrenciler$ortalama <- ogrenciler$vize * 0.4 + ogrenciler$final * 0.6
ogrenciler
```

## Filtrelemek: [satır, sütun]

Virgülden öncesi satırlar, sonrası sütunlar. Boş bırakmak "hepsi" demek.

```r
ogrenciler[ogrenciler$ortalama >= 70, ]
```

```r
ogrenciler[ogrenciler$bolum == "Bilgisayar", c("isim", "ortalama")]
```

## Sıralamak

```r
ogrenciler[order(-ogrenciler$ortalama), c("isim", "ortalama")]
```

## Gruplara göre ortalama

```r
aggregate(ortalama ~ bolum, data = ogrenciler, FUN = mean)
```

## Hızlı özet: summary

```r
summary(ogrenciler$ortalama)
```
