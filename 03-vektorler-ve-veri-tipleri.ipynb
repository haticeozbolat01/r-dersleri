## Vektör kurmak: c()

Vektör, aynı türden değerlerin sıralı bir dizisi. `c()` (combine, yani birleştir) ile kuruluyor. Atama için `<-` kullanılıyor.

```r
notlar <- c(85, 92, 78, 64, 99)
notlar
```

## Üç temel veri tipi

| R'da | Python'daki karşılığı |
|---|---|
| numeric | int, float |
| character | str |
| logical | bool |

```r
isimler <- c("Ayşe", "Mehmet", "Zeynep", "Can", "Elif")
gecti <- notlar >= 70

class(notlar)
class(isimler)
class(gecti)
gecti
```

## Vektörle işlem: döngüye gerek yok

İşlem her elemana tek tek uygulanıyor. Buna **vektörleştirilmiş işlem** deniyor.

```r
notlar + 5
notlar * 2
mean(notlar)
length(notlar)
```

## İndeksleme: R 1'den sayar

Dikkat: `notlar[-1]` son elemanı değil, **birinci hariç hepsini** verir.

```r
notlar[1]
notlar[2:4]
notlar[c(1, 5)]
notlar[-1]
```

## Mantıksal filtreleme

```r
notlar[notlar >= 80]
isimler[notlar >= 80]
sum(gecti)
```

## Vektör üretmenin kısa yolları

```r
1:10
seq(0, 100, by = 25)
rep("R", 3)
```

## Tek tip kuralı

Farklı tipleri karıştırırsan R hepsini ortak bir tipe çevirir.

```r
karisik <- c(1, "iki", TRUE)
karisik
class(karisik)
```

## Eksik değerler: NA

```r
olcumler <- c(12, NA, 15, 9)
mean(olcumler)
mean(olcumler, na.rm = TRUE)
is.na(olcumler)
```

## İsimli vektörler

```r
names(notlar) <- isimler
notlar
notlar["Zeynep"]
```
