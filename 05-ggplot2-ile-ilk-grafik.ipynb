## Kütüphaneyi yüklemek ve veriye bakmak

ggplot2 Colab'ın R ortamında kurulu geliyor.

```r
library(ggplot2)
options(repr.plot.width = 8, repr.plot.height = 5)
str(mtcars[, c("mpg", "wt", "cyl", "hp")])
```

## İlk grafik: nokta grafiği

ggplot2'de grafik katman katman kuruluyor, katmanlar `+` ile ekleniyor.

```r
ggplot(mtcars, aes(x = wt, y = mpg)) +
  geom_point()
```

## Renk eklemek

Silindir sayısı aslında bir grup; R'a bunu `factor` ile söylüyoruz. Veriye göre değişen her şey `aes`'in içine yazılır.

```r
mtcars$silindir <- factor(mtcars$cyl)

ggplot(mtcars, aes(x = wt, y = mpg, color = silindir)) +
  geom_point(size = 3)
```

## Başlık, eksen adları ve kendi renklerin

```r
renkler <- c("4" = "#F6B17A", "6" = "#C2185B", "8" = "#2B1036")

ggplot(mtcars, aes(x = wt, y = mpg, color = silindir)) +
  geom_point(size = 3) +
  scale_color_manual(values = renkler) +
  labs(
    title = "Araç ağırlaştıkça yakıt verimi düşüyor",
    subtitle = "mtcars veri seti, 32 otomobil",
    x = "Ağırlık (1000 libre)",
    y = "Galon başına mil",
    color = "Silindir"
  ) +
  theme_minimal()
```

## Çubuk grafik

```r
ggplot(mtcars, aes(x = silindir)) +
  geom_bar(fill = "#C2185B") +
  labs(title = "Kaç araçta kaç silindir var?", x = "Silindir", y = "Araç sayısı") +
  theme_minimal()

table(mtcars$silindir)
```

## Trend çizgisi eklemek

```r
ggplot(mtcars, aes(x = wt, y = mpg)) +
  geom_point() +
  geom_smooth(method = "lm", se = FALSE, color = "#C2185B") +
  theme_minimal()
```
